Time series analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# create time series data
data={
    'Date':['2025-01-01','2025-02-01','2025-03-01','2025-04-01','2025-05-01','2025-06-01','2025-07-01','2025-08-01','2025-09-01','2025-10-01','2025-11-01','2025-12-01'],
    'Sales':[120,135,128,150,160,155,170,180,175,190,200,210]
}

```

```python
# create data frame
df=pd.DataFrame(data)
```

```python
#convert date column into datetime format
df['Date']=pd.to_datetime(df['Date'])
```

```python
#set date as index
df.set_index('Date',inplace=True)
```

```python
#Display the data
print('Time Series data:')
print(df)
```

```python
#plot the time series
plt.figure(figsize=(12,6))
plt.plot(df['Sales'],marker='o',linestyle='-')
plt.title('Time Series Analysis')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```
